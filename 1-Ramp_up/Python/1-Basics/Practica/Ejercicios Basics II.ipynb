## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math

input = float(input("What's the degrees: "))
radians = input * math.pi / 180
print(f"{input}° is equal to {radians} radians")
````

## Exercise 3
Write a program that calculates the area of a parallelogram (base x height). Also with `input`

````python
base = float(input("base: "))
height = float(input("height: "))
area = (base * height) / 2
print("The area of the parallelogram is: ", area)
````

## Exercise 10
For the following string you are asked to print the following cases on screen

    "Early birds catch the worm"

1. Convert it all to uppercase
2. Convert it all to lowercase
3. Convert to uppercase only the initial of the words
4. Create a list by dividing it by its spaces
5. Replace commas `,` with semicolons `;`
6. Remove lowercase `a` letters

````python
text = "Early birds catch the worm"
print (f"uppercase: {text.upper()}")
print (f"lowercase: {text.lower()}")
print (f"upperinitial: {text.title()}")
print (f"list: {text.split()}")
print (f"replace: {text.replace(' ', ';')}")
print (f"remove a: {text.replace('a', '')}")
````
